```python
import numpy as np

from nlft_qsp.poly_matrix import MatrixPolynomial
from nlft_qsp.rand import random_polynomial

d=7        # matrix dimension
n=30       # random polynomial degree

eta=0.1    # uniform lower bound on spectrum
```

```python
# Generate random matrix polynomial

def random_matrix_polynomial(shape, N, eta):
    MP = MatrixPolynomial(shape)

    for x in range(d):
        for y in range(d):
            MP[x, y] = random_polynomial(N, eta=0)
    
    s = MP.sup_norm() # double check this method
    if s > eta:
        MP *= (1 - eta) / s
    return MP

A = random_matrix_polynomial((d, d), n+1, eta=eta) # target spectral factor

I = np.eye(d)
P = -A * A.conjugate() + I # target matrix polynomial

## Test: check if eigenvalues are bounded away from 0
# assert all(np.all(np.linalg.eigvalsh(eval) > eta) for eval in P.eval_at_roots_of_unity(N))
```

```python
from nlft_qsp.solvers.janashia_lagvilava import spectral_factor

print("n/eta = ", n/eta)

S, U = spectral_factor(P, eta, verbose=True)

N = 2 ** 10

## Test: check if S * S.conj() = P
print("SP error (analytic): \t", max(np.linalg.norm(np.array(S_eval) @ np.conjugate(np.transpose(S_eval)) - np.array(P_eval)) 
          for S_eval, P_eval in zip(S.eval_at_roots_of_unity(N), P.eval_at_roots_of_unity(N))))

# Test: check if U is unitary
print("U error: \t\t", max(np.linalg.norm(U_eval @ np.conjugate(np.transpose(U_eval)) - np.eye(d)) for U_eval in U.eval_at_roots_of_unity(N)))
```
